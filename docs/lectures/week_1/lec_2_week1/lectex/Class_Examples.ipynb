```python
measurements = [1.0, 2.0]
alias = measurements
alias.append(3.0)
# measurements is now [1.0, 2.0, 3.0]
```

```python
print(f"alias: {alias}, measurements: {measurements}") # measurements is now [1.0, 2.0, 3.0] why? because alias and measurements refer to the same list object in memory.
```

```python
independent_copy = measurements.copy() # independent copy
```

```python
print(f"measurements: {measurements}, alias: {alias}, independent_copy: {independent_copy}") # independent_copy is now [1.0, 2.0, 3.0] but it is a separate list object in memory.
```

```python
# Create two identical lists
a = [1, 2, 3]
b = [1, 2, 3]

# Check their values
print(a == b)  # True (Same contents)

# Check their memory addresses
print(a is b)  # False (Different memory)

# Make 'c' point to 'a'
c = a
print(a is c)  # True (Same exact object)
```

```python
latest_position = [1, 2, 3] # List of recent positions
altitude_m = 120 # Current altitude in meters
max_altitude_m = 100 # Maximum allowed altitude in meters


if latest_position is None: # Check if the list is empty
    status = "waiting_for_odometry" # if so, set status to "waiting_for_odometry" odemetry means the use of data from motion sensors to estimate change in position over time.
elif altitude_m > max_altitude_m: # Check if the altitude exceeds the maximum allowed altitude
    status = "limit_exceeded"
else:
    status = "ready"
```

```python
print(f"status: {status}") # Print the status
```

```python
detections = ["car", "person", "car"]
car_count = 0
for i, label in enumerate(detections):
    if label == "car" and i > 0:  # Check if the label is "car" and it's not the first detection
        car_count += 1
print(f"Number of cars detected: {car_count}")
```
